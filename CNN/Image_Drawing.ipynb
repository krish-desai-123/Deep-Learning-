```python
import cv2
import numpy as np
```

```python
kana = cv2.imread('Assets/kana.jpg')
print(f'Image shape: {kana.shape}')
```

```python
cv2.line(kana, (200,200), (500,500), (0, 255, 0), 2)

cv2.rectangle(kana, (200,200), (500,500), (255, 0, 0), 2)

cv2.circle(kana, (350,350), 100, (0,0,255), 2)

cv2.putText(kana, 'KANA', (300, 550), cv2.FONT_HERSHEY_SIMPLEX, 1, (0,0,255), 2)

cv2.imshow('Kana', kana)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
# Custom image

custom_image = np.full((500, 500, 3), (0, 0, 0), dtype=np.uint8)
print(f'Custom image shape: {custom_image.shape}')
```

```python
cv2.line(custom_image, (200, 200), (300, 300), (255, 0, 0), 2)

cv2.rectangle(custom_image, (200, 200), (400, 400), (0, 255, 0), 2)

cv2.circle(custom_image, (300, 300), 50, (0,0,255), 2)

cv2.putText(custom_image, 'Harsh', (265,390), cv2.FONT_HERSHEY_SIMPLEX, 1, (0,0,255), 2)

cv2.imshow('Custom_image', custom_image)
cv2.waitKey(0)
cv2.destroyAllWindows()
```
