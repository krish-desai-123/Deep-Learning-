```python
import cv2
```

```python
kana = cv2.imread('Assets/kana.jpg')
print(f'Image shape: {kana.shape}')
```

```python
cv2.line(kana, (200,200), (500,500), (0, 255, 0), 2)

cv2.rectangle(kana, (200,200), (500,500), (255, 0, 0), 2)

cv2.circle(kana, (350,350), 100, (0,0,255), 2)

cv2.putText(kana, 'KANA', (300, 550), cv2.FONT_HERSHEY_SIMPLEX, 1, (0,0,255), 2)


cv2.imshow('Kana', kana)
cv2.waitKey(0)
cv2.destroyAllWindows()
```
