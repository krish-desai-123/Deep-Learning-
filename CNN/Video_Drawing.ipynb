```python
import cv2
```

```python
cap = cv2.VideoCapture('Assets/meme.mp4')

width = cap.get(cv2.CAP_PROP_FRAME_WIDTH)
height = cap.get(cv2.CAP_PROP_FRAME_HEIGHT)
fps = cap.get(cv2.CAP_PROP_FPS)

print(f'FPS: {fps}')
print(f'Width: {width}')
print(f'Height: {height}')
```

```python
while True:

    ret, frame = cap.read()

    if not ret:
        break
    cv2.line(frame, (200,200), (500,300), (0, 255, 0), 2)

    cv2.rectangle(frame, (200,200), (500,300), (255, 0, 0), 2)

    cv2.circle(frame, (350,250), 50, (0,0,255), 2)

    cv2.putText(frame, 'KANA', (300, 350), cv2.FONT_HERSHEY_SIMPLEX, 1, (0,0,255), 2)
    cv2.imshow('Meme',frame)

    if cv2.waitKey(25) & 0xFF == ord('q'):
        break

cap.release()
cv2.destroyAllWindows()
```

```python
# Live Webcam

cap = cv2.VideoCapture(0)

width = cap.get(cv2.CAP_PROP_FRAME_WIDTH)
height = cap.get(cv2.CAP_PROP_FRAME_HEIGHT)
fps = cap.get(cv2.CAP_PROP_FPS)

print(f'FPS: {fps}')
print(f'Width: {width}')
print(f'Height: {height}')
```

```python
while True:

    ret, frame = cap.read()

    if not ret:
        break

    cv2.rectangle(frame, (200,100), (400,300), (255, 0, 0), 2)

    cv2.imshow('Live Webcam', frame)

    if cv2.waitKey(25) & 0xFF == ord('q'):
        break

cap.release()
cv2.destroyAllWindows()
```
