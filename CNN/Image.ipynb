```python
import cv2
import matplotlib.pyplot as plt
```

```python
kana = cv2.imread('Assets/kana.jpg')
print(f'Image shape: {kana.shape}')

cv2.imshow('Kana', kana)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
kana_plt = cv2.cvtColor(kana, cv2.COLOR_BGR2RGB)
plt.imshow(kana_plt)
plt.axis('off')
plt.show()
```

```python
# Resize

ram = cv2.imread('Assets/ram.png')
print(f'Image shape before resizing: {ram.shape}')

ram_resized = cv2.resize(ram, (500, 500))
print(f'Image shape after resizing: {ram_resized.shape}')

cv2.imwrite('Assets/ram_resized.png', ram_resized)
cv2.imshow('Ram', ram_resized)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
ram_plt = cv2.cvtColor(ram, cv2.COLOR_BGR2RGB)
plt.imshow(ram_plt)
plt.axis('off')
plt.show()
```

```python
# Grayscale

kana_gray = cv2.cvtColor(kana, cv2.COLOR_BGR2GRAY)

cv2.imwrite('Assets/kana_gray.jpg', kana_gray)
cv2.imshow('Kana_grayscale', kana_gray)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
kana_gray_plt = cv2.cvtColor(kana_gray, cv2.COLOR_GRAY2RGB)
plt.imshow(kana_gray_plt)
plt.axis('off')
plt.show()
```

```python
# GaussianBlur

kana_blurred = cv2.GaussianBlur(kana, ksize = (9, 9), sigmaX = 0 )

cv2.imwrite('Assets/kana_blurred.jpg', kana_blurred)
cv2.imshow('Kana_blurred', kana_blurred)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
kana_blurred_plt = cv2.cvtColor(kana_blurred, cv2.COLOR_BGR2RGB)
plt.imshow(kana_blurred_plt)
plt.axis('off')
plt.show()
```

```python
# Edges

kana_edge = cv2.Canny(kana, 50, 150)

cv2.imwrite('Assets/kana_edge.jpg', kana_edge)
cv2.imshow('Kana_edge', kana_edge)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
kana_edge_plt = cv2.cvtColor(kana_edge, cv2.COLOR_BGR2RGB)
plt.imshow(kana_edge_plt)
plt.axis('off')
plt.show()
```
