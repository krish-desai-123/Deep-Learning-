```python
import cv2
```

```python
# Webcam

face_cam = cv2.VideoCapture(0)

face_detector = cv2.FaceDetectorYN.create(
    'Assets/face_detection_yunet_2023mar.onnx',
    '',
    (320, 320),
    0.9,
    0.3,
    5000
)

while True:
    ret, frame = face_cam.read()

    if not ret:
        break

    x, y = frame.shape[:2]

    face_detector.setInputSize((y, x))

    _ , faces = face_detector.detect(frame)

    if faces is not None:
        for face in faces:
            x, y, w, h = face[:4].astype(int)

            cv2.rectangle(frame, (x, y), (x + w, y + h), (255, 0, 0), 2)

    cv2.imshow('Face Detection', frame)


    key = cv2.waitKey(10)

    if key == 27:
        break


face_cam.release()
cv2.destroyAllWindows()
```

```python
# Video

cap = cv2.VideoCapture('Assets/meme.mp4')

while True:

    ret, frame = cap.read()

    if not ret:
        break


    x, y = frame.shape[:2]
    face_detector.setInputSize((y, x))

    _, faces = face_detector.detect(frame)

    if faces is not None:
        for face in faces:
            x, y, w, h = face[:4].astype(int)
            cv2.rectangle(frame, (x, y), (x + w, y + h), (255, 0, 0), 2)

    cv2.imshow('Meme', frame)

    key = cv2.waitKey(10)
    if key == 27:
        break

cap.release()
cv2.destroyAllWindows()
```
