```python
import cv2
from ultralytics import YOLO

modelv12n = YOLO('Assets/yolov12n-face.pt')
```

```python
# Live Video

webcam = cv2.VideoCapture(0)

while True:
    ret, frame = webcam.read()

    if not ret:
        break

    res = modelv12n(frame, verbose = False)

    cv2.imshow('Face Detection (Live)', res[0].plot())

    if cv2.waitKey(1) & 0xFF == 27:
        break

webcam.release()
cv2.destroyAllWindows()
```

```python
# Image

image = cv2.imread('Assets/selfie.jpg')

resized_image = cv2.resize(image, dsize = None, fx = 0.5, fy = 0.5)

res = modelv12n(resized_image, verbose = False)

cv2.imshow('Face Detection (Image)', res[0].plot())

cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
# Video

video = cv2.VideoCapture('Assets/classroom.mp4')

while True:
    ret, frame = video.read()

    if not ret:
        break

    resized_frame = cv2.resize(frame, dsize = None, fx = 0.5, fy = 0.5)

    res = modelv12n(resized_frame, verbose = False)

    cv2.imshow('Face Detection (Video)', res[0].plot())

    if cv2.waitKey(25) & 0xFF == 27:
        break

video.release()
cv2.destroyAllWindows()
```
