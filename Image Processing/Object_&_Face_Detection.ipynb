```python
import cv2
from ultralytics import YOLO

model26n = YOLO('Assets/yolo26n.pt')
modelv12n = YOLO('Assets/yolov12n-face.pt')
```

```python
# Live Video

webcam = cv2.VideoCapture(0)

while True:
    ret, frame = webcam.read()

    if not ret:
        break

    res_o = model26n(frame, verbose = False)
    res_f = modelv12n(frame, verbose = False)

    res1 = res_o[0].plot()
    res = res_f[0].plot(img = res1)

    cv2.imshow('Object + Face Detection (Live)', res)

    if cv2.waitKey(1) & 0xFF == 27:
        break

webcam.release()
cv2.destroyAllWindows()
```

```python
# Image

image = cv2.imread('Assets/selfie.jpg')

resized_image = cv2.resize(image, dsize = None, fx = 0.5, fy = 0.5)

res_o = model26n(resized_image, verbose = False)
res_f = modelv12n(resized_image, verbose = False)

res1 = res_o[0].plot()
res = res_f[0].plot(img = res1)

cv2.imshow('Object + Face Detection (Image)', res)

cv2.waitKey(0)
cv2.destroyAllWindows()
```
