```python
import cv2
```

```python
# Video

cap = cv2.VideoCapture('Assets/meme.mp4')

while True:

    ret, frame = cap.read()

    if not ret:
        break

    cv2.imshow('Meme', frame)

    if cv2.waitKey(25) & 0xFF == ord('q'):
        break

cap.release()
cv2.destroyAllWindows()
```
